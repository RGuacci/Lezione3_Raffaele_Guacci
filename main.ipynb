```python
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

```python
input_text = input("Enter text: ")

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
    {
        "role" : "developer",
        "content" : (
            "Traduci il testo dall'italiano all'inglese, o dall'inglese all'italiano in base alla lingua di partenza. Restituisci solo la traduzione senza spiegazioni"
        )
     
     },
     {
        "role" : "user",
        "content" : input_text 
     }
    ],
)

print(completion.choices[0].message.content)
```
