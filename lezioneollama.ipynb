```python
import ollama
```

```python
input_text = input("Inserisci il testo: ")

response = ollama.chat(
    model = "qwen3:8b",
    messages = [
        {
        "role" : "system",
        "content" : (
            "Traduci il testo che stai per ricevere "
            "se è in italiano traducilo in inglese "
            "se è in inglese traducilo in italiano. "
            "restituisci solo il testo senza commenti"
        )
      },
      {
          "role" : "user",
          "content" : input_text
      }
    ]
)

# Ciao mi chiamo Raffaele

print(response.message.content)
```
